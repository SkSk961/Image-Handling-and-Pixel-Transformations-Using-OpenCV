15. Adjust the brightness of the image.

```python
boy_img = cv2.imread('boy.jpg')

# Create brightness adjustment matrix
matrix = np.ones(boy_img.shape, dtype='uint8') * 50

# Increase and decrease brightness
bright_img = cv2.add(boy_img, matrix)
dark_img = cv2.subtract(boy_img, matrix)

# Display images
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(cv2.cvtColor(dark_img, cv2.COLOR_BGR2RGB))
plt.title("Darker Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(cv2.cvtColor(boy_img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(cv2.cvtColor(bright_img, cv2.COLOR_BGR2RGB))
plt.title("Brighter Image")
plt.axis("off")

plt.show()
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
boy_img = cv2.imread('boy.jpg')

# Split the image into B, G, R components
B, G, R = cv2.split(boy_img)


# Show the channels.
plt.figure(figsize = [20, 10])
plt.subplot(141); plt.imshow(R); plt.title('Red Channel')
plt.subplot(142); plt.imshow(G); plt.title('Green Channel')
plt.subplot(143); plt.imshow(B); plt.title('Blue Channel')
```

21. _merged the R, G, B , display along with orginal image 

```python
# Merge the individual channels into a BGR image.
imgMerged = cv2.merge((R, G, B))
plt.figure(figsize = [20, 10])
plt.subplot(144)
plt.imshow(imgMerged)
plt.title('Merged Output');
```

22. Split the image into the H, S, V components &  Display the channels.

```python
boy_img = cv2.imread('boy.jpg')

# Convert BGR to HSV
img_hsv = cv2.cvtColor(boy_img, cv2.COLOR_BGR2HSV)

# Split HSV channels
h, s, v = cv2.split(img_hsv)

# Display the channels
plt.figure(figsize=[20, 5])

plt.subplot(141)
plt.imshow(h, cmap='gray')
plt.title('H Channel')
plt.axis('off')

plt.subplot(142)
plt.imshow(s, cmap='gray')
plt.title('S Channel')
plt.axis('off')

plt.subplot(143)
plt.imshow(v, cmap='gray')
plt.title('V Channel')
plt.axis('off')


```
